```python
def char_frequency(text):
    freq = {}
    for ch in text:
        if ch in freq:
            freq[ch] += 1
        else:
            freq[ch] = 1
    return freq
print(char_frequency("algonex"))
```

```python
def find_duplicates(transactions):
    count = {}
    duplicates = []
    for t in transactions:
        if t in count:
            count[t] += 1
        else:
            count[t] = 1
    for key in count:
        if count[key] > 1
            duplicates.append(key)
    return duplicates
print(find_duplicates([101,102,103,101,104,102]))
```

```python
def salary_report(employees):
    highest = employees[0]
    lowest = employees[0]
    total = 0
    for name, salary in employees:
        total += salary
        if salary > highest[1]:
            highest = (name, salary)
        if salary < lowest[1]:
            lowest = (name, salary)
    average = total / len(employees)
    return highest[0], lowest[0], average
employees = [
    ("John",50000),
    ("David",70000),
    ("Smith",40000),
    ("Alice",90000)
]
result = salary_report(employees)
print("Highest =", result[0])
print("Lowest =", result[1])
print("Average =", result[2])
```

```python
def transaction_report(transactions):
    credit = 0
    debit = 0
    for amount in transactions:
        if amount > 0:
            credit += amount
        else:
            debit += (amount)
    balance = credit + debit
    return credit, debit, balance
transactions = [1000,-200,500,-100,2000,-500]
result = transaction_report(transactions)
print("Total Credit =", result[0])
print("Total Debit =", result[1])
print("Final Balance =", result[2])
```
